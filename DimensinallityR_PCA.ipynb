```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
x,y=make_blobs(n_samples=500,n_features=5,centers=3,cluster_std=3,random_state=42)
```

```python
x
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
x_scaled
```

```python
# here variance is being used thats why we have data as negative after scaling
```

```python
pca=PCA(n_components=2)
x_pca=pca.fit_transform(x_scaled)
```

```python
x_pca
```

```python
df_pca=pd.DataFrame(x_pca,columns=["PC1","PC2"])
df_pca["label"]=y
```

```python
sns.scatterplot(data=df_pca,x="PC1",y="PC2",hue="label",palette="Set2")
```
